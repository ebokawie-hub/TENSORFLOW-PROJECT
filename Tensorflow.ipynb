```python
import tensorflow as tf
```

Import Libaries

```python
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt

(X_train, y_train), (X_test, y_test) = mnist.load_data()

plt.imshow(X_train[1],cmap='gray')
plt.title(y_train[1])
plt.show()
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

**Normalize** **the** **image**

```python
X_train = X_train/255.0
X_test = X_test/255.0
```

Builld The Neural Network

```python
model = tf.keras.models.Sequential([ #Sequencial builds a linear stack of layer
      tf.keras.layers.Flatten(input_shape=(28,28)),#convert image to numbers
      tf.keras.layers.Dense(128, activation='relu'),#learn pattern
      tf.keras.layers.Dense(10, activation='softmax')#predict the digit
])
```

```python
model.compile(optimizer='adam' ,#update weight and biases using gradient
              loss='sparse_categorical_crossentropy', #measures prediction
              # errror for multiclass classification with interger label
              metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10)
```

```python
model.summary()
```

```python
model.summary()
```

```python
model.evaluate(X_test,y_test)
```
